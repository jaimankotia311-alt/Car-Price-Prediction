```python
# Feature Engineering
```

```python
numeric_column = df.select_dtypes(include=["number"])
```

```python
data = df.drop(["car_ID"],axis=1)
```

```python
print(data.columns.tolist())
```

```python
catergory_column = data.select_dtypes(include=["object"])
catergory_column.head()
```

```python
from sklearn.preprocessing import LabelEncoder,OneHotEncoder, OrdinalEncoder
encoder = LabelEncoder()
```

```python
for col in data.columns:
    # print(col)
    if data[col].dtype == "str":
        # print(col)
        data[col] = encoder.fit_transform(data[[col]])
```

```python
data.head()
```

```python
import seaborn as sns

plt.figure(figsize=(17,13))
sns.heatmap(data.corr(numeric_only=True)*100 , annot=True, fmt="0.2f" )
```

```python
# Splitting Data  and model trainig
```

```python
x = data.iloc[:,:-1]
y = data["price"]
```

```python
from sklearn.model_selection import train_test_split
x_train , x_test , y_train , y_test = train_test_split(x,y, test_size=0.2 ,random_state = 42 )
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(x_train,y_train)
```

```python
print("x_train shape = ",x_train.shape)
print("x_test shape = ",x_test.shape)
print("y_train shape = ",y_train.shape)
print("x_test shape = ",x_test.shape)
```

```python
y_pred =model.predict(x_test)
```

```python
from sklearn.metrics import r2_score, mean_absolute_error , mean_squared_error , root_mean_squared_error
```

```python
# Training accracy 
r2_score(y_train,model.predict(x_train))*100
```

```python
# Testing accracy 
r2_score(y_test,y_pred)*100
```

```python
print ( "The mean absolute error for testing is : ", mean_absolute_error(y_test,y_pred))
print ( "The mean squared error for testing is : ", mean_squared_error(y_test,y_pred))  
print ( "The root mean squared error for testing is : ", root_mean_squared_error(y_test,y_pred))  
```

```python
pd.DataFrame(
    {
        "Actual Output"   : y_test,
        "Predicted Output": y_pred,
        "Bias/Error"      : y_test - y_pred,
        "Error in % ": (((y_test - y_pred)/y_test)*100 ).round(2)
    }
)
```
